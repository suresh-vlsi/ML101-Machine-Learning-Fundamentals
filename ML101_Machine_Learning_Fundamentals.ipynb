for loop has the syntax of:

for i in range (start, stop, steps):

`start` refers to the initial value of i, `stop` refers to the upper limit which is not included in loop, and `steps` refers to what will be the steps by which i is incremented after each iteration.

By default, start is considered to be from 0 and steps are considered to be increments of 1

This structure is essentially same as the structure used in c++:

for (int i = start; i < stop; i += steps)

# Classes and Object-Oriented Programming (OOP)

Python supports **object-oriented programming (OOP)**. OOP helps us organize related data and operations together.

A **class** is a blueprint. An **object** is a particular instance created from that blueprint.

For example, a circuit-design program might define a `MOSFET` class. Each MOSFET object can store its threshold voltage, width, and length, and provide methods to calculate or display its parameters.

## 1. Creating a class and an object

Use the `class` keyword to define a class. Create an object by calling the class name as if it were a function.

````python
class MOSFET:
    pass  # pass is a placeholder; the class is currently empty

# Create two objects (instances) of the MOSFET class
m1 = MOSFET()
m2 = MOSFET()

print(type(m1))
print(type(m2))
````

`m1` and `m2` are two separate objects created from the same class. The `pass` statement allows us to define an empty block temporarily.

## 2. Attributes and the `__init__()` method

**Attributes** store information belonging to an object. The special method `__init__()` initializes an object when it is created.

- `self` refers to the current object.
- `self.vth` and `self.width` are instance attributes.
- `__init__()` runs automatically when we create an object.

````python
class MOSFET:
    def __init__(self, name, vth, width, length):
        self.name = name
        self.vth = vth
        self.width = width
        self.length = length

m1 = MOSFET("M1", 0.45, 1.0, 0.18)
m2 = MOSFET("M2", 0.50, 2.0, 0.18)

print(m1.name, m1.vth, m1.width, m1.length)
print(m2.name, m2.vth, m2.width, m2.length)
````

## 3. Methods

A **method** is a function defined inside a class. Methods describe actions that an object can perform.

````python
class MOSFET:
    def __init__(self, name, vth, width, length):
        self.name = name
        self.vth = vth
        self.width = width
        self.length = length

    def display_parameters(self):
        print(f"Device: {self.name}")
        print(f"Threshold voltage: {self.vth} V")
        print(f"Width: {self.width} um")
        print(f"Length: {self.length} um")

m1 = MOSFET("M1", 0.45, 1.0, 0.18)
m1.display_parameters()
````

## 4. Updating attributes

Object attributes can be read and updated using dot notation.

````python
m1.vth = 0.48
print("Updated threshold voltage:", m1.vth)
````

## 5. Inheritance

**Inheritance** allows a new class to reuse and extend the behavior of an existing class.

The class in parentheses is the parent (base) class. The new class is the child (derived) class.

````python
class CircuitComponent:
    def __init__(self, name):
        self.name = name

    def describe(self):
        print(f"Component: {self.name}")

class Resistor(CircuitComponent):
    def __init__(self, name, resistance):
        super().__init__(name)  # initialize the parent-class part
        self.resistance = resistance

    def show_resistance(self):
        print(f"{self.name}: {self.resistance} ohms")

r1 = Resistor("R1", 1000)
r1.describe()
r1.show_resistance()
````

`super()` provides access to methods from the parent class. Here, `Resistor` inherits `describe()` from `CircuitComponent` and adds its own resistance attribute and method.

## 6. Why classes are useful in engineering projects

Classes are useful when a program must represent many entities that each have their own data and behavior.

Examples:
- **ATPG:** a `Fault` class can store a fault site, fault type, and detection status.
- **Circuit simulation:** a `Gate` class can store gate type, input nets, and output net.
- **MOSFET characterization:** a `MOSFET` class can store device dimensions and measured characteristics.
- **Machine Learning:** a model object stores learned parameters and provides methods such as `fit()` and `predict()`.

A class is not mandatory for every Python program. Use one when grouping related data and operations makes the program easier to understand and maintain.

#Step 3: Splitting the Data
One of the most important practices in machine learning is to split datasets into training and test sets.

As in, a model will **train on the training set** to learn patterns and then those patterns can be **evaluated on the test set**.

Crucially, a model should **never** see testing data during training.

This is equivalent to a student studying course materials during the semester (training set) and then testing their abilities on the following exam (testing set).

Scikit-learn provides the [`sklearn.model_selection.train_test_split`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html) method to split datasets in training and test sets.

> **Note:** A common practice to use an 80/20 or 70/30 or 75/25 split for training/testing data. There is also a third set, known as a validation set (e.g. 70/15/15 for training/validation/test) for hyperparameter tuning on but for now we'll focus on training and test sets.

````python
# 1. Define x and y
# Note the double brackets [[]] for x. scikit-learn expects x to be a 2D array.
x = df[['Expense']]
y = df['Sales']

# 2. Split the data
# test_size=0.2 means 20% of the data will be for testing, and 80% for training.
# random_state=42 is like a 'seed' which ensures that everytime the same state is used, the data will be same.
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# 3. Print the shapes to see what happened
print(f"x_train shape (rows, columns): {x_train.shape}")
print(f"y_train shape (rows): {y_train.shape}")
print(f"x_test shape (rows, columns): {x_test.shape}")
print(f"y_test shape (rows): {y_test.shape}")
````

#Choosing a Model:

Choosing a model often depends on the type of problem you're working on.

For example, there are different models that Scikit-Learn recommends whether you're working on a classification or regression problem.

We'll be using the Linear Regression Model for this example.

#Step 4: Training the Model

Since we've already loaded the dataset, now all that's left is to create an "object" of the model and `fit` the dataset to the model.

Fitting a model to a dataset involves passing it the data and asking it to figure out the patterns.

If there are labels (supervised learning), the model tries to work out the relationship between the data and the labels.

If there are no labels (unsupervised learning), the model tries to find patterns and group similar samples together.

Most Scikit-Learn models have the [`fit(x, y)`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html) method built-in, where the `x` parameter is the features and the `y` parameter is the labels.

In our case, we start by fitting a model on the training split (`x_train`, `y_train`).

````python
# Create the model "object"
model = LinearRegression()

# Train the model using our training data
# This is where the model "learns" by finding the best-fit line!
model.fit(x_train, y_train)

print("Model trained successfully!")
````

### What did the model "learn"?

The model has found the "best-fit line." A line is defined by its **slope ($m$)** and **intercept ($b$)**.

  * `y = mx + b`
  * `Sales = m * (Expense) + b`

Let's see what the model found for $m$ and $b$.

````python
# The slope (m) is stored in model.coef_
m = model.coef_[0]

# The intercept (b) is stored in model.intercept_
b = model.intercept_

print(f"The best-fit line is: Sales = {m:.2f} * (Expense) + {b:.2f}")
````

#Step 5: Evaluation

`Evaluation` refers to testing the model to see how well it performs after being trained. Metrics like `R-squared`, `Mean squared error`, `Accuracy`, `F1 score` are used to evaluate different aspects of a model and the evaluation criteria and metrics also differ from model to model.

For a Linear Regression model, `R-squared` is a common starting point as it tells us how much of the variation in the target is explained by the model. It is best read alongside an error metric such as the `Mean Squared Error` (MSE) or its square root (RMSE), which is in the same units as the target.

A model is only considered to be complete once it has been evaluated properly.

````python
# The .score() method returns the R-squared value for regression models
train_score = model.score(x_train, y_train)

print(f"R-squared on the TRAINING set: {train_score:.4f}")
print("This is measured on data the model has already seen, so it is an optimistic estimate.")
````

We should always evaluate the model on data it was **not** trained on. A score on the training data only tells us how well the model fits what it has already seen, and it can hide a problem called `overfitting`.

`Overfitting` is when a model fits the training data too closely (including its noise) and then performs poorly on new data. Flexible models, such as a very deep decision tree, can score almost perfectly on training data and still do badly on the test set. A simple one-feature linear regression like ours is unlikely to overfit, but checking on the test set is still the right habit.

````python
test_score = model.score(x_test, y_test)

# Error metrics on the test set
y_test_pred = model.predict(x_test)
mse  = mean_squared_error(y_test, y_test_pred)
rmse = mse ** 0.5

print(f"R-squared on the TEST set: {test_score:.4f}")
print(f"Test MSE:  {mse:.2f}")
print(f"Test RMSE: {rmse:.2f}  (typical prediction error, in the same units as Sales)")

# Interpret the score instead of assuming it is good
if test_score >= 0.9:
    verdict = "very good"
elif test_score >= 0.7:
    verdict = "decent"
else:
    verdict = "weak"
print(f"The model explains {test_score*100:.2f}% of the variance in Sales on unseen data, which is a {verdict} fit.")
````

Another way to actually see the accuracy of the model is to plot the `best fit line` given by the model alongside the actual data.

````python
# Use our model to predict sales for the x_test data
y_pred = model.predict(x_test)

# Sort by Expense so the line is drawn cleanly from left to right
order = np.argsort(x_test['Expense'].values)

# Plot the results!
plt.figure(figsize=(6, 6))

# Plot the ACTUAL data points (from our test set)
plt.scatter(x_test, y_test, color='blue', label='Actual Sales')

# Plot our MODEL'S PREDICTIONS (the best-fit line)
plt.plot(x_test['Expense'].values[order], y_pred[order], color='red', linewidth=2, label='Predicted Sales (Our Model)')

plt.title('Model Performance: Actual vs. Predicted', fontsize=20)
plt.xlabel('Expense', fontsize=14)
plt.ylabel('Sales', fontsize=14)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.legend()
plt.show()
````

A list comprehension creates a list from an iterable, optionally applying a transformation or condition. It is often more concise than a loop.

````python
values = [1, 2, 3, 4, 5, 6]

squares = [x ** 2 for x in values]
even_values = [x for x in values if x % 2 == 0]

print("Squares:", squares)
print("Even values:", even_values)
````

## A3. Lambda, map, filter, and reduce

A `lambda` expression creates a small anonymous function. `map()` applies a function to each item, while `filter()` keeps items that satisfy a condition. `reduce()` combines items cumulatively and is imported from `functools`.

````python
from functools import reduce

values = [1, 2, 3, 4, 5]

double_values = list(map(lambda x: 2 * x, values))
even_values = list(filter(lambda x: x % 2 == 0, values))
product = reduce(lambda a, b: a * b, values)

print(double_values)
print(even_values)
print(product)
````

# Module B – Advanced Object-Oriented Programming

## B1. Encapsulation

Encapsulation groups data and the methods that operate on it. Python uses naming conventions such as a leading underscore (`_value`) to indicate that an attribute is intended for internal use; it is not strict access control.

````python
class Fault:
    def __init__(self, site, fault_type):
        self.site = site
        self.fault_type = fault_type
        self._detected = False

    def mark_detected(self):
        self._detected = True

    def status(self):
        return "Detected" if self._detected else "Undetected"

fault = Fault("net_12", "stuck-at-0")
print(fault.status())
fault.mark_detected()
print(fault.status())
````

## B2. Polymorphism

Polymorphism allows different objects to provide methods with the same name, while implementing their behavior differently.

````python
class ANDGate:
    def evaluate(self, a, b):
        return a & b

class ORGate:
    def evaluate(self, a, b):
        return a | b

gates = [ANDGate(), ORGate()]

for gate in gates:
    print(gate.evaluate(1, 0))
````

## B3. Class variables, class methods, and static methods

An instance variable belongs to an individual object. A class variable is shared by instances through the class. A class method receives the class as `cls`; a static method does not automatically receive `self` or `cls`.

````python
class TestRun:
    tool_name = "Python ATPG Demo"
    total_runs = 0

    def __init__(self, run_id):
        self.run_id = run_id
        TestRun.total_runs += 1

    @classmethod
    def get_total_runs(cls):
        return cls.total_runs

    @staticmethod
    def is_valid_run_id(run_id):
        return isinstance(run_id, int) and run_id >= 0

run1 = TestRun(1)
run2 = TestRun(2)

print(TestRun.tool_name)
print(TestRun.get_total_runs())
print(TestRun.is_valid_run_id(2))
````

## B4. Special methods

Special methods (often called magic or dunder methods) let objects work naturally with built-in operations. For example, `__str__()` defines a readable string representation.

````python
class Net:
    def __init__(self, name, fanout):
        self.name = name
        self.fanout = fanout

    def __str__(self):
        return f"Net(name={self.name}, fanout={self.fanout})"

net = Net("clk", 12)
print(net)
````

## B5. Composition versus inheritance

Inheritance models an “is-a” relationship. Composition models a “has-a” relationship. For example, a circuit can have a collection of gates; it does not need to inherit from a gate.

````python
class Gate:
    def __init__(self, gate_type, output):
        self.gate_type = gate_type
        self.output = output

class Circuit:
    def __init__(self, name, gates):
        self.name = name
        self.gates = gates  # Circuit has gates (composition)

    def gate_count(self):
        return len(self.gates)

gates = [Gate("AND", "n1"), Gate("NOT", "n2")]
circuit = Circuit("demo", gates)
print("Gate count:", circuit.gate_count())
````

## E3. Regression metrics

Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and the coefficient of determination (R²) measure different aspects of regression performance.

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5, 0.0, 2.0, 8.0])

mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_true, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)
````

## E4. Classification metrics

For classification, common metrics include accuracy, precision, recall, and F1-score. A confusion matrix shows counts of true and predicted classes.

````python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

y_true = [0, 1, 1, 0, 1, 0]
y_pred = [0, 1, 0, 0, 1, 1]

print("Accuracy:", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1:", f1_score(y_true, y_pred))
print("Confusion matrix:\\n", confusion_matrix(y_true, y_pred))
print(classification_report(y_true, y_pred, zero_division=0))
````

## E5. Overfitting, underfitting, and generalization

- **Underfitting:** the model is too simple to capture the underlying pattern; both training and test performance may be poor.
- **Overfitting:** the model learns noise or details specific to the training data; training performance is good but test performance is worse.
- **Generalization:** how well a trained model performs on previously unseen data.

A training/test split and cross-validation help estimate generalization. Keep a final test set separate until model selection is complete.

## E6. Cross-validation

K-fold cross-validation divides the training data into K folds. The model is trained and validated repeatedly, each time using a different fold for validation.

````python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.arange(20).reshape(-1, 1)
y = 3 * X.ravel() + 2

model = LinearRegression()
scores = cross_val_score(model, X, y, cv=5, scoring="r2")

print("Fold R2 scores:", scores)
print("Mean R2:", scores.mean())
````

## F1. Bit manipulation

Bitwise operations can extract fields, set bits, clear bits, and test flags.

````python
value = 0b10110110

# Extract bits [5:2]
field = (value >> 2) & 0b1111

# Set bit 0
set_bit = value | (1 << 0)

# Clear bit 4
clear_bit = value & ~(1 << 4)

# Test bit 7
bit7_is_one = bool(value & (1 << 7))

print("Field [5:2]:", bin(field))
print("After setting bit 0:", bin(set_bit))
print("After clearing bit 4:", bin(clear_bit))
print("Bit 7 is one:", bit7_is_one)
````

## F2. Simple fault representation

A class can represent a stuck-at fault. This small example stores a fault site and stuck value; it does not implement a complete ATPG algorithm.

````python
class StuckAtFault:
    def __init__(self, net_name, stuck_value):
        if stuck_value not in (0, 1):
            raise ValueError("stuck_value must be 0 or 1")
        self.net_name = net_name
        self.stuck_value = stuck_value

    def __repr__(self):
        return f"{self.net_name} stuck-at-{self.stuck_value}"

faults = [
    StuckAtFault("n1", 0),
    StuckAtFault("n2", 1),
    StuckAtFault("n3", 0)
]

for fault in faults:
    print(fault)
````

## F3. Parsing a simple Verilog module header

Regular expressions can help extract simple text patterns. This demonstration intentionally handles only a basic ANSI-style module header; real Verilog/SystemVerilog parsing requires a proper parser because the language has complex syntax.

````python
import re

verilog_text = """
module simple_fifo (
    input logic clk,
    input logic rst_n,
    input logic [7:0] data_in,
    output logic [7:0] data_out
);
endmodule
"""

header = re.search(r"module\\s+\\w+\\s*\\((.*?)\\);", verilog_text, re.S)

if header:
    ports = re.findall(r"\\b(input|output|inout)\\b[^,]*?\\b(\\w+)\\s*(?=,|$)",
                       header.group(1))
    print("Direction and port names:")
    for direction, name in ports:
        print(direction, name)
else:
    print("Module header not found.")
````

## F5. Analyze timing or tool reports

Text processing can extract simple patterns from reports. Report formats vary by tool and version, so this is a basic illustration rather than a complete STA report parser.

````python
report_text = """
Startpoint: reg_a
Endpoint: reg_b
data arrival time     0.82
data required time    1.00
slack                  0.18
"""

import re

slack_match = re.search(r"slack\\s+(-?\\d+(?:\\.\\d+)?)", report_text)

if slack_match:
    slack = float(slack_match.group(1))
    print("Slack:", slack)
    print("Timing status:", "PASS" if slack >= 0 else "VIOLATION")
else:
    print("Slack value not found.")
````
